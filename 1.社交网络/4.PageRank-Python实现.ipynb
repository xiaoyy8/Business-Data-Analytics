## 第二部分：随机测试

```python
def PR_Gdata(N, alpha=0.8): # random > alpha, then here is a edge.
    G = np.zeros((N, N))
    for i in range(N):
        for j in range(N):
            if i == j:
                continue
            if random.random() < alpha:
                G[i][j] = 1
    return G
G1 = PR_Gdata(10)
```

```python
G1
```

```python
def pagerank_GtoM(G1, N):
    M = np.zeros((N, N))
    for i in range(N):
        D_i = sum(G1[i])
        if D_i == 0:
            continue
        for j in range(N):
            M[j][i] = G1[i][j] / D_i # 边和链接矩阵转换
    return M
M1 = pagerank_GtoM(G1, 10)
```

```python
def PageRank(M, N, T=300, eps=1e-6):
    R = np.ones(N) / N
    for time in range(T):
        R_new = np.dot(M, R)
        if np.linalg.norm(R_new - R) < eps:
            break
        R = R_new.copy()
    return R_new
```

```python
Result = PageRank(M=M1,N=10)
```

```python
Result = PageRank(M,4,T=300, eps=1e-6)
```
